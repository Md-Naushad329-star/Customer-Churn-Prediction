# 📈 Notebook 03: Exploratory Data Analysis (EDA) & Insights
### Customer Sales, Revenue & Churn Prediction Analytics System

In this notebook, we analyze customer behaviors, revenue drivers, churn risk factors, and multivariate relationships using Matplotlib and Seaborn.


```python
import os
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use('seaborn-v0_8-whitegrid')
%matplotlib inline

BASE_DIR = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
PROCESSED_DIR = BASE_DIR / 'data' / 'processed'

df = pd.read_csv(PROCESSED_DIR / 'customer_churn_cleaned.csv')
df_sales = pd.read_csv(PROCESSED_DIR / 'sales_summary.csv')
print(f'Master dataset rows: {len(df):,}')
```

## 1. Summary Statistics & High-Level KPIs


```python
total_revenue = df_sales['net_amount'].sum()
aov = df_sales['net_amount'].mean()
churn_rate = df['churn'].mean() * 100

print(f'Total Gross Revenue: ${total_revenue:,.2f}')
print(f'Average Order Value: ${aov:.2f}')
print(f'Customer Churn Rate: {churn_rate:.2f}%')

df.describe().T
```

## 2. Churn Telemetry Distributions


```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.histplot(data=df, x='satisfaction_score', hue='churn', palette={0: '#2b8a3e', 1: '#e03131'}, kde=True, ax=axes[0])
axes[0].set_title('Satisfaction Score by Churn Status')

sns.boxplot(data=df, x='churn', y='days_since_last_purchase', palette={0: '#2b8a3e', 1: '#e03131'}, ax=axes[1])
axes[1].set_xticklabels(['Retained', 'Churned'])
axes[1].set_title('Days Inactive by Churn Status')

plt.tight_layout()
plt.show()
```

## 3. Correlation Matrix Heatmap


```python
corr_cols = [
    'age', 'total_orders', 'total_spend', 'average_order_value',
    'days_since_last_purchase', 'purchase_frequency', 'support_tickets',
    'satisfaction_score', 'membership_duration_months', 'churn'
]

plt.figure(figsize=(10, 8))
sns.heatmap(df[corr_cols].corr(), annot=True, fmt='.2f', cmap='coolwarm', center=0, linewidths=0.5)
plt.title('Feature Correlation Matrix Heatmap')
plt.show()
```
